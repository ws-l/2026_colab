## 1. binary.csv를 읽어서 데이터프레임을 파악해보세요
- describe() 적용
- .columns 확인

```python
import pandas as pd
df = pd.read_csv("binary.csv")
df.describe()
print( df.columns )
df.info()
```

## 2. 결측치가 있는지 확인해보세요
- .isnull().mean() 적용
- 있다면 imputing 해보세요

```python
df.isnull().mean()
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="mean")
imputer.fit_transform( df.drop(columns="admit") )
```

## 3. binary.csv의 admit의 1,0 비율 확인하고, 1은 "pass", 0은 "fail"로 바꾸세요.

```python
df.admit.value_counts()
df.admit.replace([0,1], ["fail", "good"])
```

## 4. gre, gpa, rank에 대해서 scaling 적용하세요

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit_transform( df[["gre","gpa","rank"]] )
```
